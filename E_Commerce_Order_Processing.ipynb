```python
def process_order(price, quantity, customer_type="regular"):
   purchase = price * quantity

   if customer_type == "premium" and purchase >= 10000:
    discount = purchase * 0.20
   elif customer_type == "regular" and purchase >= 10000:
    discount = purchase * 0.10
   else:
    discount = 0

   final_amount = purchase - discount

   if final_amount >=5000:
    shipping = 0
   else:
    shipping = 250

   total = final_amount +  shipping

   print(f"Product Price: Rs. {price}")
   print(f"Quantity: {quantity}")
   print(f"Coustomer Type: {customer_type}")
   print(f"Purchase Amount: Rs.{int(purchase)}")
   print(f"Discount Amount: Rs.{int(discount)}")
   print(f"After discount Amount: Rs.{int(final_amount)}")
   print(f"Shipping fee: {shipping}")
   print(f"Total Payment is: Rs.{int(total)}")

price = int(input("Enter the product price: "))
quantity = int(input("Enter the quantity of product: "))
customer_type = input("Enter customer type (regular / premium): ")

process_order(price, quantity, customer_type)
```
